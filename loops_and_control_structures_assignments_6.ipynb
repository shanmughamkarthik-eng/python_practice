Problem 6.8: Simple Calculator (Medium)

Create a simple calculator program that:

1. Shows a menu of operations: Add, Subtract, Multiply, Divide, Exit
2. Takes two numbers as input
3. Performs the selected operation
4. Shows the result
5. Repeats until the user chooses to exit
6. Handles division by zero error

Hint: Use while loop. Display menu, get choice. Use if-elif to perform operation. Check if divisor is 0 before division.

```python
class calculator:
  def __init__(self, number_a = 0, number_b = 0):
    self.number_a = number_a
    self.number_b = number_b

  def addition(self):
    return self.number_a + self.number_b

  def subtraction(self):
    return self.number_a - self.number_b

  def multiplication(self):
      return self.number_a * self.number_b

  def division(self):
      return self.number_a / self.number_b if self.number_b != 0 else "Error: Division by zero"



print("Simple Calculator")

calc = calculator()

while True:
  print("\n1. Add\n2. Subtract\n3. Multiply\n4. Divide\n5. Exit")
  choice = input("\nEnter your choice: ")

  if choice == "5":
    print("Exiting...")
    break

  elif choice in ["1", "2", "3", "4"]:
    try:
      calc.number_a = float(input("Enter first number: "))
      calc.number_b = float(input("Enter second number: "))

      if choice == "1":
        print("Result: ", calc.addition())

      elif choice == "2":
        print("Result: ", calc.subtraction())

      elif choice == "3":
        print("Result: ", calc.multiplication())

      elif choice == "4":
        print("Result: ", calc.division())

    except ValueError:
      print("Invalid input. Please enter a valid number.")

  else:
    print("Invalid choice. Please select a valid option.")
```

##Problem 7.1: Collatz Conjecture (Medium)
The Collatz conjecture is one of the most famous unsolved problems in mathematics. The rules are:

1. Start with any positive integer n
2. If n is even, divide it by 2
3. If n is odd, multiply it by 3 and add 1
4. Repeat until you reach 1

###Write a program that:
*  Takes a number as input
*  Shows the complete sequence until it reaches 1
*  Counts how many steps it took

Example: Starting with 6: 6 → 3 → 10 → 5 → 16 → 8 → 4 → 2 → 1 (8 steps)

Hint: While n != 1: if n%2==0, n=n//2, else n=3*n+1. Count steps and print each value.

```python
print("Collatz Conjecture")

number = int(input("Enter a number: "))
steps = 0
print(f"\n{number}", end="")

while number!=1:
  if number % 2 == 0:
    number = number/2

  else:
    number = (3* number)+1

  print(f" -> {int(number)}", end="")
  steps += 1

print(f" ({steps} steps)")
```

###Problem 7.2: Caesar Cipher (Hard)
Write a program that implements a Caesar cipher to encrypt and decrypt messages.

The Caesar cipher shifts each letter by a fixed number of positions in the alphabet.

Example: With shift=3
*  'A' becomes 'D'
*  'Hello' becomes 'Khoor'

Your program should:

1. Ask for a message and shift value
2. Provide options to encrypt or decrypt
3. Handle both uppercase and lowercase letters
4. Keep non-alphabetic characters unchanged

Hint: Loop through each character. Use ord() to get ASCII value, shift it, use chr() to convert back. Use modulo 26 for wrapping. Check if char.isalpha().

```python
def get_shift_value():
  while True:
    try:
      return int(input("Enter the shift value: "))
    except ValueError:
      print("Invalid value. Please enter a shift value number.")

def caesar_cipher(message, shift_value):
  result = ""
  for char in message:
    if char.isupper():
        result += chr((ord(char) + shift_value - 65) % 26 + 65)

    elif char.islower():
        result += chr((ord(char) + shift_value - 97) % 26 + 97)

    else:
        result += char
  return result

# ------------------------

print("Caesar Cipher\n")

print("1. Encrypt")
print("2. Decrypt")
print("0. Quit")

while True:
  try:
    choice = int(input("\nEnter your choice: "))
  except ValueError:
    print("Invalid input. Please enter a valid number.")
    continue

  if choice == 0:
    print("Exiting")
    break

  elif choice == 1:
    message = input("Enter the message: ")
    shift_value = get_shift_value()
    print(f"Encrypted message: {caesar_cipher(message, shift_value)}")

  elif choice == 2:
    message = input("Enter the message: ")
    shift_value = get_shift_value()
    print(f"Decrypted message: {caesar_cipher(message, -shift_value)}")

  else:
    print("Invalid choice. Please select a valid option. eg., 0, 1 or 2")
```

```python
def withdraw_money():
  global balance
  global transaction_history

  while True:
    try:
      amount = float(input("Enter the amount: "))
    except ValueError:
      print("Please enter a numeric value.")
      continue

    if amount <= 0:
      print("Invalid amount. Please enter a positive number.")
      continue

    elif amount > balance:
      print("Insufficient balance.")
      continue

    else:
      balance -= amount
      transaction_history.append(f" - {amount}")
      print(f"Withdrawal successful. New balance: {balance}")
      return

#----------------------------------------------------------------

def deposit_money():
  global balance
  global transaction_history

  while True:
    try:
      amount = float(input("Enter the amount: "))
    except ValueError:
      print("Please enter a numeric value.")
      continue

    if amount <= 0:
      print("Invalid amount. Please enter a positive number.")
      continue

    else:
      balance += amount
      transaction_history.append(f" + {amount}")
      print(f"Deposit successful. New balance: {balance}")
      return

#-----------------------------------------------------------------

def check_balance():
  global balance
  global transaction_history

  print(f"Your balance is: {balance:,.2f}")

  if input("Do you want to see the last 5 transactions? (y/n): ").lower() == 'y':
     # This slices the list to get the last 5 items
    last_five = transaction_history[-5:]

    if not last_five:
      print("No transactions yet.")

    else:
      print("\n--- Recent Transactions ---")
      for tx in last_five:
        print(tx)
      return

  else:
    return

#-------------------------------------------------------------------

def change_pin():
  global pin
  global transaction_history
  attempts = 0

  while attempts < 3:
    try:
      current = int(input(f"Enter your current PIN (Attempt {attempts + 1}/3):"))

      if current != pin:
          print("Incorrect PIN. Authorization failed.")
          attempts += 1
          continue

      else:
        try:
          new_pin = int(input("Enter the new PIN: "))
          confirmation = int(input("Confirm the new PIN: "))

          if new_pin == confirmation:
            pin = new_pin
            print("PIN changed successfully.")
            return True

          else:
            print("Pin not matched")
            return False

        except ValueError:
          print("Please enter a numeric value.")
          continue

    except ValueError:
      print("Please enter a numeric value.")
      attempts += 1
      continue

  if attempts >= 3:
    print("Maximum number of attempts reached. Exiting.")
    return False

#-------------------------------------------------------------------------
def login():
  global pin
  attempts = 0

  while attempts < 3:
    try:
      entered_pin = int(input(f"Enter your PIN  (Attempt {attempts + 1}/3): "))
      if entered_pin == pin:
        print("Login successful.")
        return True

      else:
          print("Incorrect PIN. Authorization failed.")
          attempts += 1
          continue

    except ValueError:
      print("Please enter a numeric value.")
      attempts += 1
      continue

  return False

#-----------------------------------------------

print("ATM Machine Simulator")
balance = 10000
pin = 1234
transaction_history = []

if not login():
  print("Exiting")

else:
  while True:
    print("\n1. Check Balance")
    print("2. Withdraw Money")
    print("3. Deposit Money")
    print("4. Change PIN")
    print("5. Exit")

    try:
      choice = int(input("\nEnter your choice: "))
    except ValueError:
      print("Invalid input. Please enter a valid number.")
      continue

    if choice == 5:
      print("Exiting")
      break

    elif choice == 1:
      check_balance()

    elif choice == 2:
      withdraw_money()

    elif choice == 3:
      deposit_money()

    elif choice == 4:
      if not change_pin():
        break

    else: print("Invalid choice. Please select a valid option. eg., 1 to 5")
```

####Problem 8.15: Traffic Light System (Easy)
Based on light color, print action: Red=Stop, Yellow=Caution, Green=Go

Hint: Use if-elif-else or dictionary to map color to action.

```python
import random
import time

def get_random_color(last_color):
    colors = ["Red", "Yellow", "Green"]

    if last_color in colors:
        colors.remove(last_color)
    return random.choice(colors)


print("Traffic Light system")
last_light = None

try:
    while True:
        light_color = get_random_color(last_light)
        last_light = light_color

        if light_color == "Red":
            print(f"{light_color} signal. Stop")
            time.sleep(3)

        elif light_color == "Yellow":
            print(f"{light_color} signal. Caution")
            time.sleep(1)

        elif light_color == "Green":
            print(f"{light_color} signal. Go")
            time.sleep(3)

except KeyboardInterrupt:
    print("Program terminated by user")
```

Hint: Calculate discriminant = b**2 - 4ac. Use if-elif-else to check if > 0, == 0, or < 0.

Calculate employee bonus:
1. Salary < 10000: 10% bonus
2. Salary 10000-20000: 15% bonus
3. Salary > 20000: 20% bonus
4. If years of service > 5: Add extra 5%

Hint: Determine bonus percentage based on salary.

If years > 5, add 5 to bonus percentage.

Calculate: bonus = salary * bonus_percentage / 100.

```python
print("Bonus Calculation")
try:
    salary = float(input("Enter the salary: "))
    years_of_service = int(input("Enter the years of service: "))

    if salary < 0:
        print("Invalid input. Please enter a non-negative salary")

    else:
        if salary < 10000:
            bonus_percentage = 10

        elif salary <= 20000:
            bonus_percentage = 15

        else:
            bonus_percentage = 20

        if years_of_service <= 0:
            print("Invalid input. Please enter a positive number of years of service")

        else:
            if years_of_service > 5:
                bonus_percentage += 5

            bonus = salary * bonus_percentage / 100
            print(f"Total Bonus Percentage: {bonus_percentage}%")
            print(f"Bonus Amount: Rs.{bonus:.2f}")
            print(f"Total Pay: Rs.{(salary + bonus):.2f}")

except ValueError:
    print("Invalid input. Please enter a valid number")
```

####Problem 8.24: Water State (Easy)
Determine water state based on temperature:
1. Below 0°C: Ice
2. 0-100°C: Water
3. Above 100°C: Steam

Hint: If temp < 0: Ice, elif temp <= 100: Water, else: Steam.

```python
print("Water State")
try:
    temperature = float(input("Enter the temperature in degrees Celsius: "))

    if temperature < 0:
        print("Water state is : Ice")

    elif temperature <= 100:
        print("Water state is : Water")

    else:
        print("Water state is : Steam")

except ValueError:
    print("Invalid input. Please enter a valid number")
```

####Problem 8.25: Alphabet or Digit (Easy)
Check if a character is an alphabet or a digit.

Hint: Use ch.isalpha() to check alphabet, ch.isdigit() for digit.

```python
print("Alphabet or Digit")

character = input("Enter a character: ")
try:
    if len(character) != 1:
        print("Invalid input. Please enter a single character")

    else:
        if character.isalpha():
            print(f"'{character}' is an alphabet")

        elif character.isdigit():
            print(f"'{character}' is a digit")

        else:
            print(f"'{character}' is not an alphabet or a digit")
            print("Please enter an alphabet or a digit")

except ValueError:
    print("Invalid input. Please enter a valid")
```

Problem 8.26: Tax Calculator (Medium)
Calculate income tax:

1. Income 0-250000: No tax
2. Income 250001-500000: 5% tax
3. Income 500001-1000000: 10% tax
4. Income above 1000000: 20% tax

Hint: Use if-elif-else to determine tax rate. Calculate: tax = income * tax_rate / 100.

```python
print("Tax Calculator")
try:
    income = float(input("Enter the income: "))

except ValueError:
    print("Invalid input. Please enter a valid number")

if income < 0:
    print("Invalid input. Please enter a non-negative income")

else:
    if income <= 250000:
        tax_rate = 0

    elif income <= 500000:
            tax_rate = 5

    elif income <= 1000000:
            tax_rate = 10

    else:
            tax_rate = 20

    tax = income * tax_rate / 100

    print(f"Income Tax for {income} is : Rs.{tax:.2f}")
```

Problem 6.8: Simple Calculator (Medium)

Create a simple calculator program that:

1. Shows a menu of operations: Add, Subtract, Multiply, Divide, Exit
2. Takes two numbers as input
3. Performs the selected operation
4. Shows the result
5. Repeats until the user chooses to exit
6. Handles division by zero error

Hint: Use while loop. Display menu, get choice. Use if-elif to perform operation. Check if divisor is 0 before division.

```python
class calculator:
  def __init__(self, number_a = 0, number_b = 0):
    self.number_a = number_a
    self.number_b = number_b

  def addition(self):
    return self.number_a + self.number_b

  def subtraction(self):
    return self.number_a - self.number_b

  def multiplication(self):
      return self.number_a * self.number_b

  def division(self):
      return self.number_a / self.number_b if self.number_b != 0 else "Error: Division by zero"



print("Simple Calculator")

calc = calculator()

while True:
  print("\n1. Add\n2. Subtract\n3. Multiply\n4. Divide\n5. Exit")
  choice = input("\nEnter your choice: ")

  if choice == "5":
    print("Exiting...")
    break

  elif choice in ["1", "2", "3", "4"]:
    try:
      calc.number_a = float(input("Enter first number: "))
      calc.number_b = float(input("Enter second number: "))

      if choice == "1":
        print("Result: ", calc.addition())

      elif choice == "2":
        print("Result: ", calc.subtraction())

      elif choice == "3":
        print("Result: ", calc.multiplication())

      elif choice == "4":
        print("Result: ", calc.division())

    except ValueError:
      print("Invalid input. Please enter a valid number.")

  else:
    print("Invalid choice. Please select a valid option.")
```
